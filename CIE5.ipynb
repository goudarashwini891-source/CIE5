```python
from nltk.util import ngrams
from nltk.tokenize import wordpunct_tokenize

# Given sentence
sentence = """AI is simulation of human inteiligence by machine.
It includes expects system NLP speech recognition and machine vision"""

# Tokenize the sentence
words = wordpunct_tokenize(sentence)

# Generate bigrams
bigrams = list(ngrams(words, 2))

# Generate trigrams
trigrams = list(ngrams(words, 3))

# Display the results
print("Original Sentence:")
print(sentence)

print("\nBigrams:")
for bigram in bigrams:
    print(bigram)

print("\nTrigrams:")
for trigram in trigrams:
    print(trigram)
```
